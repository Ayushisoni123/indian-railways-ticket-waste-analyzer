```python
import pandas as pd
# Load Dataset
df = pd.read_csv("indian_railway_passenger_chart.csv")

# Basic exploration
print("Dataset Shape :",df.shape)
print("\nColumns :", df.columns.tolist())
print("\nMissing values\n", df.isnull().sum())
print("\nBasic stats :\n", df.describe())
```

```python
# Key metrics summary
total_journeys = len(df)
avg_occupancy = df['Occupancy_Percentage'].mean()
avg_empty_seats = df['Available_Seats'].mean()
avg_waitlisted = df['Waitlisted_Passengers'].mean()
avg_cancellations = df['Cancellation_Count'].mean()

print("=" * 50)
print("KEY FINDINGS SUMMARY")
print("=" * 50)
print(f"Total journeys analyzed: {total_journeys}")
print(f"Average occupancy: {avg_occupancy:.1f}%")
print(f"Average empty seats per journey: {avg_empty_seats:.1f}")
print(f"Average waitlisted passengers: {avg_waitlisted:.1f}")
print(f"Average cancellations per journey: {avg_cancellations:.1f}")
print(f"\nEstimated seats wasted per journey: {avg_empty_seats:.1f}")
# NOTE: 13,000 trains/day is from Indian Railways Ministry publications
# This is an external figure used to scale our sample findings nationally
# Source: Indian Railways Year Book 2024-25
print(f"Estimated annual waste (approx 13,940 trains/day):")
print(f"  {avg_empty_seats:.0f} seats x 13,940 trains = {avg_empty_seats * 13000:,.0f} empty seats daily")
# 13,940 daily passenger trains — Source: Indian Railways Year Book 2024-25, Ministry of Railways, Train Utilisation section
```

## Finding 4 — Waitlisted Passengers vs Available Seats

The scatter plot reveals two distinct patterns:

Pattern 1 (Top Left): High waitlisted passengers with zero available seats - these are high demand routes where confirmed
capacity is fully allocated. These are prime candidates for a transfer mechanism if confirmed passengers do not wish to travel and does not cancel their tickets due to poor refund policy.

Pattern 2 (Bottom Right): High available seats with zero
waitlisted passengers — these are low demand routes where
seats go unbooked entirely. This requires a different
intervention (dynamic pricing or route marketing) rather
than a transfer mechanism. Notably some of these may be
essential rural connectivity routes where railways maintains
service despite low commercial demand — shutting them is not
viable as they serve as the only transport infrastructure
for remote communities.

Limitation: This dataset does not contain (no-show data
directly) any specific or related columns that can show the ghost occupancy . Ghost occupancy inference requires external
cancellation revenue data from Indian Railways publications.

## Dataset Limitation
Synthetic dataset — 63-64% in this dataset.
Real Railways premium train occupancy averages 86.3%  this gap cannot be justified using synthetic dataset.
Source: PIB Release Feb 2020 —
https://www.pib.gov.in/newsite/printrelease.aspx?relid=149606&reg=48&lang=2

This synthetic dataset limitation reinforces why real IRCTC
data access would be required for production-level analysis.
The framework and methodology demonstrated here remain valid
— only the input data would change.

```python
# Filter journeys where cancellations(cancellation count) are above median
# and waitlisted passengers are above zero
# print(df['Cancellation_Count'].median())
high_cancel = df['Cancellation_Count'] > df['Cancellation_Count'].median()
# print(high_cancel)
has_waitlist = df['Waitlisted_Passengers']>0
# print(has_waitlist)
target_routes = df[high_cancel & has_waitlist]
# print(target_routes)
print(f"Total journeys analyzed: {len(df)}")
print(f"Journeys with high cancellations AND waitlited passengers: {len(target_routes)}")
print(f"Percentage of journeys that need transfer mechanism: {len(target_routes)/len(df)*100:.1f}%")
print(f"\nTop zones affected :")
print(target_routes['Zone'].value_counts())
print(f"\nTop trains affected :")
print(target_routes['Train_Type'].value_counts())
```



## Finding 8 — Target Routes For Transfer Mechanism

90 journeys (9%) show simultaneous high cancellations
AND waitlisted passengers — these are prime candidates
for a passenger ticket transfer mechanism.

Policy Recommendation:
Pilot the transfer mechanism
  Targeting:
- SR Zone first (17 affected journeys shows highest impact)
- Express train category (32 affected journeys —
  it has widest reach,high connectivity, serves all passenger demographics)
  A phased pilot approach is recommended — beginning with
SR Zone Express trains before national rollout — to test
system load, passenger behaviour, and refund processing
at manageable scale before full implementation.

Mechanism: Non-travelling confirmed passengers can  flag
unused tickets before chart preparation and the  Waitlisted passengers and those needing last-minute travel can book these flagged seats . These flagged ticket passengers pay full fare and  Original passenger receives
90% refund from this fare and rest 10% can be  retained as platform processing fee.

Conclusion :
9% identified in this 1000-record synthetic dataset.
At real IRCTC scale of 14.5 lakh daily ticket bookings (IRCTC Annual Report 2025-26 daily transactions), even
5% would represent 72,500+ journeys on daily basis where a transfer mechanism would prevent seat waste and will enable the effective utilization of seats.
